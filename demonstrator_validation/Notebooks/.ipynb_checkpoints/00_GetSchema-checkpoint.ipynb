# Obtein the data schema from the comunity for Creep experiments.

In this case we have the data structure available as an excel sheet. We are going to process it to make read it into a structured python dictionary and save it as a JSON file, which allows us to read it in an ordered way.

The file is in an excel sheet available in a google drive for development. In the future this will be availbale as an ontology with public access.

```python
schema_file= "../../Data Schema/2024-09_Schema_IUC02_v1.1.json"
```

```python
with open(schema_file, 'r') as f:
    schema = json.load(f)
```

```python
JSON(schema)
```

```python
if not os.path.exists(schema_file):
    source_url = 'https://docs.google.com/uc?export=download&id=1ujqPXifOg9dB6qBTKpwbzECfpttiHkCP'
    os.makedirs(os.path.dirname(schema_file), exist_ok=True)
    import gdown
    gdown.download(source_url, schema_file, quiet=False)
```

we can translate this development version into a JSON for navigation and reading

```python
SU = SchemaUpdater(
    schema_file, sheet_name='2023-26-05', header=6, 
)
```

```python
JSON(SU.schema_creep_test)
```
