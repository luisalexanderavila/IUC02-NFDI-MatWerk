```python
import networkx as nx
from rdflib import Graph
from pyvis.network import Network

# Load SHACL file
shacl_graph = Graph()
shacl_graph.parse("../Data/BAMDataset/Vh5205_C-98_translated.ttl", format="turtle")

# Create a directed graph
G = nx.DiGraph()

# Add edges based on SHACL constraints
for subj, pred, obj in shacl_graph:
    G.add_edge(str(subj), str(obj), label=str(pred))

# Identify root nodes (nodes with no incoming edges)
root_nodes = {node for node in G.nodes if G.in_degree(node) == 0}

# Create Pyvis Network
net = Network(notebook=True, directed=True, height="600px", width="100%")

# Add nodes and edges to Pyvis with hierarchical settings
for node in G.nodes:
    hidden = node not in root_nodes  # Hide non-root nodes initially
    net.add_node(node, label=node, title=node, hidden=hidden, physics=False)

for u, v, d in G.edges(data=True):
    net.add_edge(u, v, title=d["label"], hidden=True)  # Initially hidden edges

# Configure hierarchical layout & expand on click
net.toggle_physics(False)  # Fix positioning
net.set_options("""
var options = {
  "layout": {
    "hierarchical": {
      "enabled": true,
      "direction": "UD",
      "sortMethod": "directed"
    }
  },
  "interaction": {
    "hover": true,
    "navigationButtons": true
  },
  "physics": {
    "enabled": false
  }
}
""")

# Save and display
net.show("shacl_graph.html")
```

```python
import networkx as nx
from rdflib import Graph
from pyvis.network import Network

# Load SHACL file
shacl_graph = Graph()
shacl_graph.parse("../Data/BAMDataset/Vh5205_C-99_translated.json", format="turtle")

# Create NetworkX graph
G = nx.DiGraph()

# Add edges based on SHACL constraints
for subj, pred, obj in shacl_graph:
    G.add_edge(str(subj), str(obj), label=str(pred))

# Identify root nodes (nodes that never appear as an object)
all_nodes = set(G.nodes)
child_nodes = {v for _, v in G.edges}
root_nodes = all_nodes - child_nodes  # Nodes that are only subjects, never objects

# Create an interactive Pyvis network
net = Network(notebook=True, directed=True)

# Add nodes with expand/collapse functionality
for node in G.nodes:
    is_root = node in root_nodes
    net.add_node(
        node,
        label=node.split("#")[-1],  # Show only the last part of the URI
        title=node,  # Full URI on hover
        shape="ellipse",
        physics=is_root,  # Keep only root nodes expanded
        hidden=not is_root,  # Hide child nodes initially
    )

# Add edges
for u, v, d in G.edges(data=True):
    net.add_edge(u, v, title=d["label"], arrows="to")

# Configure hierarchical layout for tree structure
net.options.layout = {
    "hierarchical": {
        "enabled": True,
        "direction": "UD",  # Top to Bottom
        "sortMethod": "directed"
    }
}

# Save and display the interactive visualization
net.show("shacl_graph.html")

# Show in Jupyter Notebook
from IPython.display import display, HTML
display(HTML("shacl_graph.html"))
```

```python
import networkx as nx
import matplotlib.pyplot as plt
from rdflib import Graph, URIRef

# Load SHACL shapes
g = Graph()
g.parse("../Data/BAMDataset/Vh5205_C-78_translated.ttl", format="turtle")

# Create a directed graph
G = nx.DiGraph()

# Add nodes and edges based on SHACL predicates
for subj, pred, obj in g:
    G.add_edge(str(subj), str(obj), label=str(pred))

# Draw the graph
plt.figure(figsize=(10, 5))
pos = nx.spring_layout(G)
nx.draw(G, pos, with_labels=True, node_color="lightblue", edge_color="gray", node_size=2000, font_size=8)
edge_labels = {(u, v): d["label"] for u, v, d in G.edges(data=True)}
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)
plt.show()
```
